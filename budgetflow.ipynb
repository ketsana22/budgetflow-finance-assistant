# Budget Calculator

```python
def calculate_budget(income, expenses):
    total_expenses = sum(expenses)
    remaining_balance = income - total_expenses

    return total_expenses, remaining_balance


# Example financial data
monthly_income = 3500
monthly_expenses = [600, 350, 400, 800]

total, remaining = calculate_budget(
    monthly_income, monthly_expenses
)

print("Total expenses: $", total)
print("Remaining balance: $", remaining)
```

# A spending summary by catergory

```python

def spending_summary(spending_dict):
    summary = {}

    for category, amounts in spending_dict.items():
        if isinstance(amounts, list):
            summary[category] = sum(amounts)
        else:
            summary[category] = amounts

    return summary


monthly_spending_data = {
    "Rent": [800],
    "Food": [600],
    "Transport": [300, 50],
    "Entertainment": [150, 250]
}

summary = spending_summary(monthly_spending_data)

print("Spending Summary:")
for category, amount in summary.items():
    print(f"{category}: ${amount}")

print("Total Spending: $", sum(summary.values()))
```

# input validation

```python

def validate_financial_values(income, expenses):
    """Check that income and expenses contain valid values."""

    if not isinstance(income, (int, float)) or isinstance(income, bool):
        raise ValueError("Income must be a number.")

    if income < 0:
        raise ValueError("Income cannot be negative.")

    if isinstance(expenses, dict):
        for category, amount in expenses.items():
            if not isinstance(category, str) or not category.strip():
                raise ValueError("Each expense must have a category name.")

            if not isinstance(amount, (int, float)) or isinstance(amount, bool):
                raise ValueError(f"{category} must be a number.")

            if amount < 0:
                raise ValueError(f"{category} cannot be negative.")

    elif isinstance(expenses, list):
        for amount in expenses:
            if not isinstance(amount, (int, float)) or isinstance(amount, bool):
                raise ValueError("Each expense must be a number.")

            if amount < 0:
                raise ValueError("Expenses cannot be negative.")

    else:
        raise ValueError("Expenses must be a list or dictionary.")

    return True


# test with valid financial values
try:
    income = 3500
    expenses = {"Rent": 800, "Food": 600, "Transport": 350, "Entertainment": 400}
    validate_financial_values(income, expenses)
    print("All financial values are valid")
except ValueError as e:
    print(f"Error: {e}")
```

#Budget assistant

```python

def budget_assistant(income, expenses):
    # Validate the financial values first
    validate_financial_values(income, expenses)

    # Calculate total expenses
    if isinstance(expenses, dict):
        total_expenses = sum(expenses.values())
    else:
        total_expenses = sum(expenses)

    # Calculate remaining balance
    remaining_balance = income - total_expenses

    # Generate budget advice
    print("BudgetFlow Assistant")
    print("--------------------")

    if remaining_balance < 0:
        print("You are spending more than your income.")
        print("Try reducing non-essential expenses.")
    elif remaining_balance == 0:
        print("You have used all your income.")
        print("Try saving some money next month.")
    else:
        print("Your expenses are within your income.")
        print(f"You have ${remaining_balance:.2f} remaining.")

        if remaining_balance >= income * 0.20:
            print("Good job! You have at least 20% of your income left.")
        else:
            print("Try saving more by reviewing your expenses.")

    print("--------------------")


# Example
monthly_income = 3500

monthly_spending = {
    "Rent": 800,
    "Food": 600,
    "Transport": 350,
    "Entertainment": 400
}

budget_assistant(monthly_income, monthly_spending)
```

```python

import pandas as pd

def analyse_transactions(file):
    if file is None:
        return "Please upload a CSV file."

    try:
        df = pd.read_csv(file.name)

        required_columns = ["Category", "Amount"]

        if not all(column in df.columns for column in required_columns):
            return "Your CSV must contain 'Category' and 'Amount' columns."

        df["Amount"] = pd.to_numeric(df["Amount"], errors="coerce")

        if df["Category"].isna().any() or df["Amount"].isna().any():
            return "Please check your file for missing or invalid categories or amounts."

        if (df["Amount"] < 0).any():
            return "Transaction amounts cannot be negative."

        summary = df.groupby("Category")["Amount"].sum()
        total = df["Amount"].sum()

        result = "Spending by category:\n"
        for category, amount in summary.items():
            result += f"{category}: ${amount:.2f}\n"

        result += f"\nTotal transactions: {len(df)}"
        result += f"\nTotal spending: ${total:.2f}"

        return result

    except Exception as error:
        return f"Could not read the file: {error}"

print("CSV transaction analysis is ready.")
```

```python

import gradio as gr
import pandas as pd

def csv_upload_app(file):
    if file is None:
        return "Please upload a CSV file."

    try:
        df = pd.read_csv(file)

        required_columns = ["Category", "Amount"]

        if not all(column in df.columns for column in required_columns):
            return "Your CSV must contain Category and Amount columns."

        df["Amount"] = pd.to_numeric(df["Amount"], errors="coerce")

        if df["Category"].isna().any() or df["Amount"].isna().any():
            return "Please check your CSV for missing or invalid values."

        if (df["Amount"] < 0).any():
            return "Transaction amounts cannot be negative."

        summary = df.groupby("Category")["Amount"].sum()
        total = df["Amount"].sum()

        result = "Spending by category:\n"

        for category, amount in summary.items():
            result += f"{category}: ${amount:.2f}\n"

        result += f"\nTotal transactions: {len(df)}"
        result += f"\nTotal spending: ${total:.2f}"

        return result

    except Exception:
        return "Unable to read the file. Please upload a valid CSV."

csv_demo = gr.Interface(
    fn=csv_upload_app,
    inputs=gr.File(
        label="Upload Transaction CSV",
        file_types=[".csv"],
        type="filepath"
    ),
    outputs=gr.Textbox(label="Spending Analysis"),
    title="BudgetFlow - CSV Spending Analysis",
    description="Upload a CSV containing Category and Amount columns."
)

csv_demo.launch()
```

```python

import gradio as gr

def budgetflow_app(income, rent, food, transport, entertainment):
    try:
        values = [income, rent, food, transport, entertainment]

        if any(value is None for value in values):
            return "Please enter a value in every field."

        if any(value < 0 for value in values):
            return "Income and expenses cannot be negative."

        expenses = {
            "Rent": rent,
            "Food": food,
            "Transport": transport,
            "Entertainment": entertainment
        }

        total_expenses = sum(expenses.values())
        remaining = income - total_expenses

        breakdown = "\n".join(
            f"{category}: ${amount:.2f}"
            for category, amount in expenses.items()
        )

        return (
            f"Monthly Spending Summary\n\n{breakdown}\n\n"
            f"Total expenses: ${total_expenses:.2f}\n"
            f"Remaining balance: ${remaining:.2f}"
        )

    except (TypeError, ValueError):
        return "Please enter valid numbers."

demo_categories = gr.Interface(
    fn=budgetflow_app,
    inputs=[
        gr.Number(label="Monthly Income ($)", minimum=0),
        gr.Number(label="Rent ($)", minimum=0),
        gr.Number(label="Food ($)", minimum=0),
        gr.Number(label="Transport ($)", minimum=0),
        gr.Number(label="Entertainment ($)", minimum=0)
    ],
    outputs=gr.Textbox(label="Budget Results"),
    title="BudgetFlow - Personal Finance Assistant",
    description="Enter your income and monthly expenses by category."
)

demo_categories.launch()
```
